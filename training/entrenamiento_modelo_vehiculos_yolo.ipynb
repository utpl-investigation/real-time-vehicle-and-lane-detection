```python
from roboflow import Roboflow
#Insertar API de roboflow
rf = Roboflow(api_key="")
#Insertar id del workspace, colocar id del project
project = rf.workspace("").project("")
#Insertar la versión del project, sustituir por el 2
version = project.version(2)
#insertar el modelo que se va a utilizar
dataset = version.download("yolov8")


```

```python
from ultralytics import YOLO
#colocar la versión de YOLO a utilizar
model = YOLO("yolo11m-obb.pt")
```

```python
data_path = "/content/datasetV2-2/data.yaml"
results = model.train(data=data_path,
                          epochs=300,
                          patience=50,
                          imgsz=800,           # Clave 1: Reduce memoria
                          batch=16,            # Clave 2: Mantiene batch
                          amp=True,            # Clave 3: Precisión mixta
                          dropout=0.3,         # Clave 4: Regularización
                          mosaic=0.5,          # Clave 5: Reduce carga
                          plots=True)          # Clave 6: permiter ver resultados
```
